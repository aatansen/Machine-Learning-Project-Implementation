### Import Libraries

```python
import pandas as pd
from evidently import Report
from evidently.presets import DataDriftPreset,DataSummaryPreset
from sklearn.model_selection import train_test_split
```

### Load Dataset

```python
# Load the Visa dataset
data_url = "../data/raw/EasyVisa.csv"
visa_frame = pd.read_csv(data_url)
```

### Train Test Split

```python
reference_data, current_data = train_test_split(
    visa_frame,
    test_size=0.2,   # 20% for current_data
    random_state=42  # for reproducibility
)
```

```python

reference_data.shape
```

```python
current_data.shape
```

### Data Drift report

```python
# Create a Data Drift report
visa_data_drift_report = Report(metrics=[DataDriftPreset(),DataSummaryPreset()])
```

```python
# Run the report
my_report=visa_data_drift_report.run(reference_data=reference_data, current_data=current_data)
```

### Testing function code (works only on `DataDriftPreset` and used in `data_validation`)
*This code used in `data_validation` using only `DataDriftPreset` it will work properly but after adding `DataSummaryPreset` it won't work properly*

```python
import json
report_json_str = my_report.json()
json_report = json.loads(report_json_str)
# Extract drift info
n_drifted_features = json_report['metrics'][0]['value']
n_features = len(json_report['metrics']) - 1

metric_id = json_report['metrics'][0]['metric_id']
drift_share = n_drifted_features['share']
evidently_threshold = float(metric_id.split("drift_share=")[1].strip(")"))

drift_status = drift_share >= evidently_threshold

print(f"{int(n_drifted_features['count'])}/{n_features} features drifted "
                        f"(drift share: {drift_share:.2%}, threshold: {evidently_threshold})")
```

### Testing function code (works on both `DataDriftPreset,DataSummaryPreset`)

```python
import json

report_json_str = my_report.json()
json_report = json.loads(report_json_str)
metrics = json_report.get("metrics", [])

# --- find DriftedColumnsCount metric ---
drift_metric = next(
    (m for m in metrics if m["metric_id"].startswith("DriftedColumnsCount")),
    None,
)
if not drift_metric:
    raise ValueError("DriftedColumnsCount metric missing!")

drift_count = drift_metric["value"]["count"]
drift_share = drift_metric["value"]["share"]

# --- threshold from metric_id ---
threshold = float(
    drift_metric["metric_id"].split("drift_share=")[1].strip(")")
)

# --- total number of columns ---
col_metric = next(
    (m for m in metrics if m["metric_id"] == "ColumnCount()"),
    None,
)

if col_metric:                     # ✅ summary available
    n_columns = int(col_metric["value"])
else:                              # 🚩 summary removed → count ValueDrift entries
    n_columns = sum(
        1 for m in metrics if m["metric_id"].startswith("ValueDrift(")
    )

drift_status = drift_share >= threshold

print(f"{int(drift_count)}/{n_columns} features drifted "
    f"(drift share: {drift_share:.2%}, threshold: {threshold})")
```
